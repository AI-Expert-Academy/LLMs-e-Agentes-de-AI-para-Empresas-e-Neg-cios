## Ajuste de exibição do resultado

````python
def format_res(res, return_thinking=False):
  res = res.strip()

  if return_thinking:
    res = res.replace("<think>", "[pensando...] ")
    res = res.replace("</think>", "\n---\n")

  else:
    if "</think>" in res:
      res = res.split("</think>")[-1].strip()

  return res

def show_res(res):
    from IPython.display import Markdown
    display(Markdown(res))
````

````python
res = format_res(output.text, return_thinking=False)
show_res(res)
````

### Saída estruturada completa (opcional)

Como sugestão alternativa, você também poderia pedir para a LLM retornar uma saída estruturada ("structured output"), por exemplo no formato json. Isso foi abordado no projeto para departamento de Recursos Humanos, consulte caso queira saber mais detalhes como.

Para isso, basta fornecer explicitamente no prompt como você deseja que seja retornada a resposta. Por exemplo, adicionar assim:

```
FORMATO DE RESPOSTA:
   **Resumo Executivo**:
   [1-2 frases com os principais achados]

   **Dados Mensais**:
   | Mês | Entradas (R$) | Saídas (R$) | Lucro Líquido (R$) |
   |-----|---------------|-------------|--------------------|
   [preencher com dados extraídos]

   **Principais Insights**:
   - Padrão 1: [descrição + impacto]
   - Padrão 2: [descrição + impacto]
   
   **Recomendações**:
   - [Sugestão 1 baseada nos dados]
   - [Sugestão 2 baseada nos dados]
```

E o LlamaIndex também possui outros métodos próprios para auxiliar a geração de saída estruturada. Para mais detalhes, veja: https://docs.llamaindex.ai/en/stable/examples/structured_outputs/structured_outputs/


#### Teste de saída estruturada

````python
template_saida_estruturada = """
Você é um analista financeiro.
Ao ler o relatório a seguir, extraia insights financeiros relevantes e explique-os de forma clara e didática, como se estivesse apresentando para gestores não especialistas. 
Utilize linguagem acessível e destaque pontos importantes sobre lucros, despesas, fluxo de caixa, riscos e oportunidades.
Retorne a saída em formato estruturado no seguinte formato:

**Resumo Executivo**:
[1-2 frases com os principais achados]

**Dados Mensais**:
| Mês | Entradas (R$) | Saídas (R$) | Lucro Líquido (R$) |
|-----|---------------|-------------|--------------------|
[preencher com dados extraídos]

**Principais Insights**:
- Padrão 1: [descrição + impacto]
- Padrão 2: [descrição + impacto]

**Recomendações**:
- [Sugestão 1 baseada nos dados]
- [Sugestão 2 baseada nos dados]

Caso você não consiga formular uma recomendação devido a falta de dados, deixe claro que não é possível fornecer uma recomendação específica.

Conteúdo do documento:
{}
"""

full_prompt_saida_estruturada = template_saida_estruturada.format(content)

output_se = llm.complete(full_prompt_saida_estruturada)
#print(output_se.text)
res_se = format_res(output_se.text)
show_res(res_se)
````

## Limitando o tamanho (opcional)

Muitas vezes o tamanho do prompt enviado ao modelo estoura a capacidade máximo de tokens. Para casos assim é possível limitar o tamanho do conteúdo enviado, por exemplo, usando apenas os primeiros 500 caracteres do conteúdo, por exemplo, ou utilizando RAG.

Fazer uma limitação de tamanho por caracteres é simples e pode ser útil para casos onde você saiba qual o conteúdo que deve ser enviado, mas e para casos onde você não sabe qual parte do conteúdo é mais relevante? Para isso, a técnica de RAG (retrieval-augmented generation) ganha espaço devido a busca semântica inerente ao seu processo.
