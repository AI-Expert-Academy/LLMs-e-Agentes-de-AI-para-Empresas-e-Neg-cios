## Ajuste de exibição do resultado

````python
def format_res(res, return_thinking=False):
  res = res.strip()

  if return_thinking:
    res = res.replace("<think>", "[pensando...] ")
    res = res.replace("</think>", "\n---\n")

  else:
    if "</think>" in res:
      res = res.split("</think>")[-1].strip()

  return res

def show_res(res):
    from IPython.display import Markdown
    display(Markdown(res))
````

````python
res = format_res(output.text, return_thinking=False)
show_res(res)
````

### Saída estruturada completa (opcional)

Como sugestão alternativa, você também poderia pedir para a LLM retornar uma saída estruturada ("structured output"), por exemplo no formato json. Isso foi abordado no projeto para departamento de Recursos Humanos, consulte caso queira saber mais detalhes como.

Para isso, basta fornecer explicitamente no prompt como você deseja que seja retornada a resposta. Por exemplo, adicionar assim:

```
FORMATO DE RESPOSTA:
   **Resumo Executivo**:
   [1-2 frases com os principais achados]

   **Dados Mensais**:
   | Mês | Entradas (R$) | Saídas (R$) | Lucro Líquido (R$) |
   |-----|---------------|-------------|--------------------|
   [preencher com dados extraídos]

   **Principais Insights**:
   - Padrão 1: [descrição + impacto]
   - Padrão 2: [descrição + impacto]
   
   **Recomendações**:
   - [Sugestão 1 baseada nos dados]
   - [Sugestão 2 baseada nos dados]
```

E o LlamaIndex também possui outros métodos próprios para auxiliar a geração de saída estruturada. Para mais detalhes, veja: https://docs.llamaindex.ai/en/stable/examples/structured_outputs/structured_outputs/

# Uso com agente (ReAct) + RAG

````python
from llama_index.core.tools import BaseTool, FunctionTool
from llama_index.core.agent.workflow import ReActAgent, AgentStream, ToolCallResult
from llama_index.core.workflow import Context
````

## Criação de tools de exemplo

````python
def multiply(a: float, b: float) -> float:
    """Multiply two numbers and return the result.."""
    return a * b

def divide(a: float, b: float) -> float:
    """Divide two numbers and return the result."""
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b

def add(a: float, b: float) -> float:
    """Add two numbers and return the result."""
    return a + b

def subtract(a: float, b: float) -> float:
    """Subtract two numbers and return the result."""
    return a - b

multiply_tool = FunctionTool.from_defaults(fn=multiply, name="multiply")
add_tool = FunctionTool.from_defaults(fn=add)
divide_tool = FunctionTool.from_defaults(fn=divide)
subtract_tool = FunctionTool.from_defaults(fn=subtract)
````

````python
from llama_index.llms.groq import Groq

# deepseek-r1-distill-llama-70b descomissionado durante o curso
llm = Groq(model="qwen/qwen3.8-27b", temperature=0.1)

from llama_index.core import Settings
Settings.llm = llm
````

````python
tools = [multiply_tool, add_tool, divide_tool, subtract_tool]

agent = ReActAgent(tools=tools, llm=llm, verbose=True, kwargs={"tool_choice": "auto"})
ctx = Context(agent)
````

## Geração do resultado com agente

````python
handler = agent.run(user_msg="Qual é o resultado de 20+(2*4)?", ctx=ctx)

async for ev in handler.stream_events():
    # if isinstance(ev, ToolCallResult):
    #     print(f"\nCall {ev.tool_name} with {ev.tool_kwargs}\nReturned: {ev.tool_output}")
    if isinstance(ev, AgentStream):
        print(f"{ev.delta}", end="", flush=True)

response = await handler
````

````python
print(str(response))
````

````python
print(response.tool_calls)
````

### Um exemplo mais complexo

Este é um exemplo mais complexo de uso de agente ReAct. Exemplo usando operações aritméticas são absorvidas pelo próprio modelo devido sua facilidade, mas algo que seja mais inerente a natureza humana como uma piada fica mais fácil de ser resolvido com o uso de uma tool pois o modelo não é treinado para esse tipo de tarefa.

````python
def pave_ou_pacume() -> str:
    """Return a string indicating if it's pavê or pacumê."""
    return "É pavê ou pacumê?"

mario = FunctionTool.from_defaults(fn=pave_ou_pacume, name="pave_ou_pacume")
````

````python
from llama_index.core.memory import Memory

agent_mario = ReActAgent(tools=[pave_ou_pacume], llm=llm, verbose=True)
ctx_mario = Context(agent_mario)
memory = Memory.from_defaults()

handler = agent_mario.run(user_msg="É pavê?", ctx=ctx_mario, memory=memory)
````

````python
async for ev in handler.stream_events():
    if isinstance(ev, ToolCallResult):
        print(f"\nCall {ev.tool_name} with {ev.tool_kwargs}\nReturned: {ev.tool_output}")
    if isinstance(ev, AgentStream):
        print(f"{ev.delta}", end="", flush=True)

response = await handler
````

````python
print(str(response))
print(response.tool_calls)
````

## Consultas com Query Engine

````python
query_engine = index.as_query_engine(similarity_top_k=3, llm=llm)
````

````python
from llama_index.core.tools import QueryEngineTool, ToolMetadata

def create_query_engine(llm, index):
    query_engine = index.as_query_engine(similarity_top_k=3, llm=llm)
    query_engine_tool = QueryEngineTool(
        query_engine=query_engine,
        metadata=ToolMetadata(
            name="doc_search",
            description=("Provides information about the company's finances.")
        )
    )
    return query_engine_tool
````

````python
query_engine_tool = create_query_engine(llm, index)
````

## Criação do agente

````python
from llama_index.core.agent.workflow import FunctionAgent

agent = FunctionAgent(tools=[query_engine_tool], llm=llm, verbose=True)

handler = agent.run(user_msg="Qual foi o total de despesas operacionais em 2024?")

async for event in handler.stream_events():
    if isinstance(event, ToolCallResult):
        print(
            f"Call {event.tool_name} with args {event.tool_kwargs}\nReturned: {event.tool_output}"
        )
    elif isinstance(event, AgentStream):
        print(event.delta, end="", flush=True)

response = await handler
````

````python
print("Resposta do agente:")
print(str(response))
print("\n####################\n")
print("Tools usadas")
print(response.tool_calls)
````

````python
def translate(text, source_lang="pt", target_lang="en", llm=None):
    from deep_translator import GoogleTranslator
    if llm is None:
        res = GoogleTranslator(source=source_lang, target=target_lang).translate(text)
    else:
        prompt = f"Traduza o seguinte texto de {source_lang} para {target_lang} (retorne somente o texto e mais nenhuma informação): '{text}'"
        res = llm.complete(prompt).text.strip()
        res = format_res(res)
    return res
````

## Expandindo as capacidades do agente com tools

````python
def growth_rate(previous: float, current: float) -> float:
    """Useful for calculating the percentage growth rate between two values."""
    return ((current - previous) / previous) * 100

growth_rate_tool = FunctionTool.from_defaults(fn=growth_rate)
````

````python
tools = [query_engine_tool, growth_rate_tool]

agent = ReActAgent(tools=tools, llm=llm, verbose=True)
````

````python
query = "compare as despesas operacionais de 2024 e 2023 e calcule a taxa de crescimento"
# query = "qual foi o total de despesas operacionais da empresa em 2024?"

ipt = translate(query, llm=llm)
ctx = Context(agent)
handler = agent.run(user_msg=ipt, ctx=ctx)

async for ev in handler.stream_events():
    if isinstance(ev, ToolCallResult):
        print(f"\nCall {ev.tool_name} with {ev.tool_kwargs}\nReturned: {ev.tool_output}")
    if isinstance(ev, AgentStream):
        print(f"{ev.delta}", end="", flush=True)

response = await handler
````

````python
print("Resposta do agente:")
print(translate(str(response), "en", "pt", llm=llm))
print("\n####################\n")
print("Tools usadas")
print(response.tool_calls)
````
